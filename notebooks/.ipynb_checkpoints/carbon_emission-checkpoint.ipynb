```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data = pd.read_csv('./dataset/Carbon_(CO2)_Emissions_by_Country.csv') 
```

```python
mada = data[data["Country"] == "Madagascar"]
```

```python
data.duplicated().sum()
data.isna().sum()
```

```python
countries = pd.get_dummies(data["Country"])
countries = countries.astype(int)

date = pd.get_dummies(data["Date"])
date = date.astype(int)

region = pd.get_dummies(data['Region'])
region = region.astype(int)

data.join([countries,date,region])
```

```python
data = data.drop(['Country','Date','Region'] , inplace = True , axis = 1) 
```
