## Creating a dataframe 
A dataFrame column is a Series datastructure 
1- To create a dataFrame : use python regular dict type 
2 - the describe() method provides a quick overview of the numerical data in a DataFrame 
and returns a Series . 

```python
df = pd.DataFrame({
    "age" : [40,75,72], 
    "name" : ["Kim" , "Donald" , "Xi"],
    "country" : ["North Korea" , "USA" , "China"]
})

x = zip(df["country"], df["name"], df["age"])
for e in x : 
    print(e)
```

```python
df.describe()
```

A pandas Series is 1-dimensional and only the number of rows is returned.
A pandas DataFrame is multidimensional and the number of rows and columns are returned 

```python
df.age.shape
df[["age", "country"]].shape
```
