# VECTORS

```python
def vector_add(v , w) : 
    return [v_i + w_i for v_i , w_i in zip(v,w)]
def vector_substract(v,w) : 
    return [v_i + w_i for v_i , w_i in zip(v,w)] 
    
# example : 
v = [1,1]
```

```python
#sums all corresponding elementsvector_add(v,w)
def vector_sum(*vectors):
    res = vectors[0]
    for vector in vectors[1:]:
        res = vector_add(res , vector) 
    return res

vector_sum([1,2]  ,[1,2])
```

```python
def scalar_multiply(c,w) : 
    return [c * i for i in w]
```

```python
def vector_mean(*vectors) : 
    n = len(vectors) 
    res = scalar_multiply(1/n , vector_sum(vectors))
    return res
vector_mean([1,2],[3,4])
```

The dot product measures how far the vector v extends in the w direction. For example, if
w = [1, 0] then dot(v, w) is just the first component of v. Another way of saying this is
that it’s the length of the vector you’d get if you projected v onto w

```python
def dot(v,w) : 
    return sum([v_i * w_i for v_i , w_i in zip(v,w)])
    
def sum_of_square(v) : 
    return dot(v,v)

#length of a vector v 
import math
def magnitude(v) : 
    return math.sqrt(sum_of_square(v))

def distance(v, w):
    return magnitude(vector_subtract(v, w))
```
