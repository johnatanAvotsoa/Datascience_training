### dataset sample : Is a customer likely to buy a car given the following features : 
### they have children , their income level and have they already have a house 

```python
df = pd.DataFrame([
    [1 ,'high'    , 0 , 1] , 
    [0 ,'low'     , 1 , 0] , 
    [1, 'medium'  , 0 , 0] , 
    [0 ,'high'    , 0 , 1] , 
    [1 ,'low'     , 1 , 0] , 
    [0 ,'medium' ,  1 , 1] , 
    [1 ,'high'   ,  1 , 1]
] , columns = ['child' , 'income' , 'has-house' , 'buy-car'] )

features = df.drop(columns=['buy-car'])

n = len(df)
```

```python
import math

def entropy(series) :
    probability = series.value_counts(normalize=True)# normalize = True for computing probability
    return - np.sum(probability * np.log2(probability))
```

Compute each feature (child , income , has-house) information gain to decide which one will be the tree root node : 
it will be the one with the highest info gain

```python
def info_gain(target,features,attribute) : 
    total_entropy = entropy(df[target])
    s = 0 
    for value in features[attribute].unique() : 
        d = df[target].loc[features[attribute] == value]
        #l = d.groupby(a)[s].value_counts()
        #print(value)
        s += (len(d) / n ) * entropy(d) 
    return total_entropy - s
```
